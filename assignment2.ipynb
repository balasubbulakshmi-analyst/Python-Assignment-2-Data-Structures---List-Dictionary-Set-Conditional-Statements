```python
# 1.LIST CREATION
# a. age list with five integers
# b. name list with five strings
age_list=[24,25,26,27,28]
name_list=["abi","uma","bala","raju","ravi"]
print("age_list:",age_list)
print("name_list",name_list)
```

```python
# 2.List Operations / Modifications:
# a. Append the string "Yazhini" to name_list.
name_list.append("yazhini")
print("after append", name_list)
```

```python
# b.Insert the element 30 at index 2 in age_list.
age_list.insert(2,30)
print("after insert", age_list)
```

```python
# c.Remove the string "Yazhini" from name_list.
name_list.remove("yazhini")
print("after remove", name_list)
```

```python
# d. Pop the last element from age_list.
popped = age_list.pop()
print("pop element:", popped)
```

```python
# e.Extend the age_list with additional ages [29, 30, 26].
age_list.extend([29,30,26])
print("after extend:", age_list)
```

```python
# f. Sort age_list in descending order.
age_list.sort(reverse=True)
print("descending order", age_list)
```

```python
# Find Max age, Min age and sum of all ages from age_list.

print("max age:", max(age_list))

print("min age:", min(age_list))

print("sum of age:", sum(age_list))
```

```python
# 3. Accessing List Elements:
# a. Print the first element of name_list.
print("first element:", name_list[0])

# b.Print the last element of name_list.
print("last element:", name_list[-1])

#c.Print the elements from index 2 to index 4 in name_list.
print("index 2 to 4", name_list[2:5])

# d. Print the elements of name_list in reverse order.
print("reverse name list:", name_list[::-1])
```

```python
#Sets (Operations):
#a. Create a set called my_set with following values:    ['a','e','i','o','u','a','a','i']
my_set=set(['a','e','i','o','u','a','a','i'])
print(my_set)
```

Explanation: A set stores only unique elements, so the duplicate values 'a' (3 times) and 'i' (2 times) are removed automatically. The output contains just the 5 vowels: a, e, i, o, u. Also, sets are unordered, so the elements may be printed in any order, and the order can change between runs.

```python
#Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.
my_set[4]='s'
```

Explanation: This raises TypeError: 'set' object does not support item assignment. Sets are unordered and have no index positions, so we cannot access or assign elements using my_set[4]. To change a set, use add(), remove(), discard() or update() instead.

```python
# c. Create two sets:
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("set1", set1)
print("set2",set2)
```
